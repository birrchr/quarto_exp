```python
#| echo: false
import sys
sys.path.insert(0, "docs/assets")
from portal_helpers import load, fig_defaults, kpi_row, data_note, COLORS
import plotly.graph_objects as go
import plotly.express as px
import pandas as pd

data = load("chronic")
```

```python
#| echo: false
overall = sorted(data["overall"], key=lambda r: r["prevalence"], reverse=True)
labels  = [r["condition"] for r in overall]
vals    = [r["prevalence"] for r in overall]
bar_colors = [
    COLORS["danger"] if v > 20 else
    COLORS["warning"] if v > 10 else
    COLORS["primary"]
    for v in vals
]

fig = go.Figure(go.Bar(
    x=vals, y=labels,
    orientation="h",
    marker_color=bar_colors,
    text=[f"{v}%" for v in vals],
    textposition="outside",
    hovertemplate="%{y}: %{x}%<extra></extra>",
))
fig.update_layout(
    xaxis_title="Prevalence (% of respondents)",
    yaxis=dict(autorange="reversed"),
)
fig_defaults(fig, "Chronic Condition Prevalence — All Ages", height=420).show()
```

```python
#| echo: false
by_age_df = pd.DataFrame(data["by_age"])
conditions = by_age_df["condition"].unique().tolist()
age_groups = sorted(by_age_df["age_group"].unique().tolist())

# Build one trace per condition, only first is visible
fig = go.Figure()
for i, cond in enumerate(conditions):
    subset = by_age_df[by_age_df["condition"] == cond]
    vals = []
    for ag in age_groups:
        row = subset[subset["age_group"] == ag]
        vals.append(float(row["prevalence"].values[0]) if len(row) else 0)
    fig.add_trace(go.Bar(
        name=cond,
        x=age_groups,
        y=vals,
        marker_color=COLORS["primary"],
        visible=(i == 0),
        hovertemplate=f"{cond}: %{{y:.1f}}%<extra></extra>",
    ))

# Dropdown menu
buttons = [
    dict(
        label=cond,
        method="update",
        args=[
            {"visible": [j == i for j in range(len(conditions))]},
            {"title": f"{cond} — Prevalence by Age Group"},
        ],
    )
    for i, cond in enumerate(conditions)
]

fig.update_layout(
    updatemenus=[dict(
        type="dropdown",
        direction="down",
        x=0.01, xanchor="left", y=1.15,
        showactive=True,
        buttons=buttons,
    )],
    yaxis_title="Prevalence (%)",
    xaxis_title="Age group",
)
fig_defaults(fig, f"{conditions[0]} — Prevalence by Age Group", height=430).show()
```

```python
#| echo: false
multi = pd.DataFrame(data["multimorbidity"])
fig = go.Figure(go.Bar(
    x=multi["chronic_count"].astype(str) + " condition(s)",
    y=multi["pct"],
    marker_color=[
        COLORS["success"] if n == 0 else
        COLORS["primary"] if n == 1 else
        COLORS["warning"] if n <= 3 else
        COLORS["danger"]
        for n in multi["chronic_count"]
    ],
    text=[f"{p}%" for p in multi["pct"]],
    textposition="outside",
    hovertemplate="Count: %{x}, %{y:.1f}% of respondents<extra></extra>",
))
fig.update_layout(yaxis_title="% of respondents", xaxis_title="Number of chronic conditions")
fig_defaults(fig, "Distribution of Multimorbidity", height=380).show()
```
